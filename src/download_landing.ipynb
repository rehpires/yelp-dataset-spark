```python
import os
import shutil

import kagglehub

from config import DATASETS, landing_file_path, source_filename
```

```python
missing = [dataset for dataset in DATASETS if not os.path.exists(landing_file_path(dataset))]

if not missing:
    print("Todos os arquivos ja estao na camada landing, nada a baixar.")
else:
    print(f"Faltando na landing: {missing}. Baixando dataset do Kaggle...")
    download_dir = kagglehub.dataset_download("yelp-dataset/yelp-dataset")

    for dataset in missing:
        source_path = os.path.join(download_dir, source_filename(dataset))
        destination_path = landing_file_path(dataset)
        shutil.copy(source_path, destination_path)
        print(f"{dataset}: {source_path} -> {destination_path}")
```

```python
for dataset in DATASETS:
    exists = os.path.exists(landing_file_path(dataset))
    print(f"{dataset}: {'OK' if exists else 'FALTANDO'} ({landing_file_path(dataset)})")
```
