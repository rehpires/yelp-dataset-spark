```python
from pyspark.sql import functions as F

from config import BRONZE_SCHEMAS, DATASETS, bronze_table, landing_volume_dir
```

```python
for dataset in DATASETS:
    df = (
        spark.read.schema(BRONZE_SCHEMAS[dataset])
        .json(landing_volume_dir(dataset))
        .withColumn("_source_file", F.col("_metadata.file_path"))
        .withColumn("_ingested_at", F.current_timestamp())
    )

    df.write.format("delta").mode("overwrite").saveAsTable(bronze_table(dataset))

    print(f"{dataset}: {df.count()} linhas gravadas em {bronze_table(dataset)}")
```
